• **Tipos incorretos:** datas lidas como texto, códigos numéricos armazenados como strings, colunas numéricas com separador inadequado.

• **Inconsistências entre tabelas:** chaves secundárias sem correspondência (por exemplo, um id_pedido existente em **itens_pedido** mas ausente na tabela **pedidos**).

• **Registros inválidos:** datas invertidas, pagamentos com valores negativos ou fretes fora da realidade.

```python
# Resumo Detalhado das Tabelas (incluindo Duplicatas por ID)

resumo = []

for nome, df in tabelas.items():
    # 1. Identifica colunas que provavelmente são IDs (contêm 'id' ou 'customer', 'order', etc)
    cols_id = [c for c in df.columns if 'id' in c.lower() or 'pk' in c.lower() or c.lower().startswith('customer') or c.lower().startswith('order') or c.lower().startswith('seller') or c.lower().startswith('product')]

    # 2. Calcula duplicatas na chave principal (se existir)
    dup_chave = 0
    chave_usada = "N/A"
    if cols_id:
        # Prioriza IDs que parecem ser chave primária da própria tabela
        pk_candidates = [c for c in cols_id if nome[:-1] + '_id' in c.lower() or nome + '_id' in c.lower() or c.lower() == 'review_id' or c.lower() == 'order_id' or c.lower() == 'customer_id' or c.lower() == 'seller_id' or c.lower() == 'product_id']

        if pk_candidates:
            chave_usada = pk_candidates[0]
            dup_chave = df.duplicated(subset=[chave_usada]).sum()
        elif cols_id:
            chave_usada = cols_id[0] # Usa o primeiro ID encontrado como fallback
            dup_chave = df.duplicated(subset=[chave_usada]).sum()

    resumo.append({
        "Tabela":       nome,
        "Linhas":       df.shape[0],
        "Colunas":      df.shape[1],
        "Nulos Total":  df.isnull().sum().sum(),
        "Nulos %":      f"{(df.isnull().sum().sum() / (df.size) * 100):.1f}%",
        "Dup. Linha":   df.duplicated().sum(),
        "Dup. ID":      dup_chave,
        "Chave Usada":  chave_usada
    })

df_resumo = pd.DataFrame(resumo)
print("Resumo detalhado das tabelas:")
display(df_resumo)
```

```python
# Tratamento e Combinação de Texto de Avaliações
original_total_reviews = len(df_avaliacoes)

def combine_review_text(row):
    title = row['review_comment_title']
    message = row['review_comment_message']

    # Convert to string and handle NaN as empty string
    title_str = str(title) if pd.notna(title) else ''
    message_str = str(message) if pd.notna(message) else ''

    if title_str and message_str:
        return f"{title_str} - {message_str}"
    elif title_str:
        return title_str
    elif message_str:
        return message_str
    else:
        return ''

df_avaliacoes['review_text_combined'] = df_avaliacoes.apply(combine_review_text, axis=1)

# Filter df_avaliacoes to remove rows where 'review_text_combined' is empty
df_avaliacoes = df_avaliacoes[df_avaliacoes['review_text_combined'] != ''].copy()

print(f"Total de avaliações originais: {original_total_reviews:,}")
print(f"Avaliações com texto combinado (título ou mensagem): {len(df_avaliacoes):,}")
print(f"Avaliações removidas (sem título e sem comentário original): {original_total_reviews - len(df_avaliacoes):,}")
```
